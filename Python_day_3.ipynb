#### While Loop

Keep doing this while the condition is true.

```python
count = 1

while count <= 5:
    print(count)
    count = count + 1
```

count = count + 1

Break : to stop the loop immediately

```python
for number in range(1, 10):
    if number == 5:
        break
    print(number)
```

### continue 
 continue - skip this iteration and continue with the next one 

```python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
```

```python
for number in range(1, 8):
    if number == 3:
        print("hi")
        continue
        print("hello")
    print(number)
```

Exercise 5 
<pre>
Given confidence_scores = [0.95, 0.72, 0.88, 0.41, 0.91]

Loop through the scores and print:

0.95 → High confidence
0.72 → Medium confidence
0.88 → High confidence
0.41 → Low confidence
0.91 → High confide</pre>nce




```python
conf = [0.95, 0.72, 0.88, 0.41, 0.91]
num = len(conf)
for i in range(num):
    if conf[i]>=0.80 and conf[i]<=1.0:
        print(f"{conf[i]} -> High confidence")
    elif conf[i]>=0.50 and conf[i]<=0.79:
        print(f"{conf[i]} -> Medium confidence")
    elif conf[i]>=0 and conf[i]<=0.5:
        print(f"{conf[i]} -> low confidence")
    else:
        print(f"{conf[i]} -> Invalid range")
```
